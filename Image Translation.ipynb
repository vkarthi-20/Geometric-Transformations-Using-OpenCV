# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image using OpenCV
img=cv2.imread("baseball.jpg", 1)
```

```python
M=np.float32([[1,0,50],
              [0,1,80]])
```

```python
# Display the image using Matplotlib
translated=cv2.warpAffine(img,M,(500,500))
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB))
plt.title("Translated Image")
plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
```

```python
M1 = np.float32([[1, 0.5, 0],
                 [0, 1, 0]])
```

```python
horizontal = cv2.warpAffine(img, M1, (img.shape[1], img.shape[0]))
```

```python
M2 = np.float32([[1, 0, 0],
                 [0.5, 1, 0]])
```

```python
vertical = cv2.warpAffine(img, M2, (img.shape[1], img.shape[0]))
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(horizontal)
plt.title("Horizontally Sheared Image")

plt.subplot(1, 3, 3)
plt.imshow(vertical)
plt.title("Vertically Sheared Image")

plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
```

```python
horizontal = cv2.flip(img, 1)
```

```python
vertical = cv2.flip(img, 0)
```

```python
both = cv2.flip(img, -1)
```

```python
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(horizontal)
plt.title("Horizontal Reflection")

plt.subplot(2, 2, 3)
plt.imshow(vertical)
plt.title("Vertical Reflection")

plt.subplot(2, 2, 4)
plt.imshow(both)
plt.title("Both-axis Reflection")

plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
```

```python
M1 = cv2.getRotationMatrix2D((img.shape[1]/2, img.shape[0]/2), 45, 1)
```

```python
rotated45 = cv2.warpAffine(img, M1, (img.shape[1],img.shape[0]))
```

```python
M2 = cv2.getRotationMatrix2D((img.shape[1]/2, img.shape[0]/2), 90, 1)
```

```python
rotated90 = cv2.warpAffine(img, M2, (img.shape[1], img.shape[0]))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(rotated45)
plt.title("45° Rotated Image")

plt.subplot(1, 3, 3)
plt.imshow(rotated90)
plt.title("90° Rotated Image")

plt.show()
```
